```python
import numpy as np
import scipy

x = np.linspace(0,1,101)
def f(x):
    return np.exp(-x)
def A(x):
    return 1-np.exp(-x)
def B(x):
    return np.exp(-x)-np.exp(-1)

import matplotlib.pyplot as plt

plt.plot(x,f(x),label='f(x)')
plt.legend()
plt.show()

num_A = scipy.integrate.cumulative_simpson(f(x), x=x, initial=0)
num_B = scipy.integrate.cumulative_simpson(f(x)[::-1], x=x, initial=0)[::-1]

plt.plot(x,num_A,label='scipy A')
plt.plot(x,num_B,label='scipy B')
plt.plot(x,A(x),label='anal A', ls='--')
plt.plot(x,B(x),label='anal B', ls='--')
plt.legend()
plt.show()

```
